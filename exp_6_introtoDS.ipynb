```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Loading the dataset
df = pd.read_csv("/content/titanic_dataset.csv")

# Previewing data
print(df.head())
```

```python
# Simple Line Plot
x = [1, 2, 3, 4, 5]
y = [3, 6, 2, 7, 1]
plt.figure(figsize=(6, 4))
sns.lineplot(x=x, y=y)
plt.title('Line Plot')
plt.show()

# Multi-line Plot
y1 = [3, 5, 2, 6, 1]
y2 = [1, 6, 4, 3, 8]
y3 = [5, 2, 7, 1, 4]
plt.figure(figsize=(6, 4))
sns.lineplot(x=x, y=y1, label='Line 1')
sns.lineplot(x=x, y=y2, label='Line 2')
sns.lineplot(x=x, y=y3, label='Line 3')
plt.title('Multi Line Plot')
plt.show()
```

```python
# Bar Chart: Average Fare per Embarkation Point
plt.figure(figsize=(8, 5))
sns.barplot(x='Embarked', y='Fare', data=df, palette='rainbow')
plt.title("Fare Of Passenger By Embarked Town")
plt.show()

# Scatter Plot: Age vs Fare
plt.figure(figsize=(8, 5))
sns.scatterplot(x="Age", y="Fare", data=df)
plt.title('Scatterplot of Age vs Fare')
plt.show()

# Bubble Chart: Age vs Fare segmented by Class
plt.figure(figsize=(8, 5))
sns.scatterplot(x="Age", y="Fare", size="Pclass", data=df, sizes=(30, 200), hue="Pclass")
plt.title('Bubble Chart of Age vs Fare, Size by Passenger Class')
plt.show()
```

```python
# Histogram: Class Distribution with Survival Hue
plt.figure(figsize=(8, 5))
sns.histplot(data=df, x="Pclass", hue="Survived", kde=True, multiple="stack")
plt.title("Survival Count by Passenger Class")
plt.show()

# Box Plot: Age distribution across Classes
plt.figure(figsize=(8, 5))
sns.boxplot(x='Pclass', y='Age', data=df, palette='rainbow')
plt.title("Age By Passenger Class")
plt.show()

# Violin Plot: Fare distribution per Class
plt.figure(figsize=(8, 5))
sns.violinplot(x="Pclass", y="Fare", data=df)
plt.title('Violin Plot of Fare by Passenger Class')
plt.show()

# Density Plot: Overall Age Distribution
plt.figure(figsize=(8, 5))
sns.kdeplot(data=df['Age'], fill=True)
plt.title('Density Plot of Passenger Ages')
plt.show()

# Heatmap of Numerical Correlations
plt.figure(figsize=(10, 8))
numeric_df = df.select_dtypes(include=['float64', 'int64'])
corr_matrix = numeric_df.corr()
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm')
plt.title('Heatmap of Titanic Dataset')
plt.show()
```
